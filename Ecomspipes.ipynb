In customers dataframe,customer_id is the primary key.In order_items dataframe,order_item_id is the primary key,order_id and product_id are the foreign keys as these keys are present in orders and in products dataframe.In orders dataframe,order_id is the primary key while customer_id is the foreign key in orders dataframe because it is present in customers dataframe.In Payments dataframe payment_id is the primary key and order_id is the foreign key of payments dataframe because it is present in order_items and in orders dataframe while in products dataframe,product_id is the primary key

Q9) Standardize inconsistent categorical values such as "completed" to "Completed" and "upi" to "UPI"

```python
#9
import pandas as pd
df3=pd.read_csv("orders.csv")
df4=pd.read_csv("payments.csv")
df3["order_status"]=df3["order_status"].str.replace("completed","Completed")
df4["payment_method"]=df4["payment_method"].str.replace("upi","UPI")
print(df3) # Dataframe of orders
print(df4) # Dataframe of payments

print(df3["order_status"]) # Values of order status dataframe
print(df4["payment_method"]) # Values of payment method dataframe
```

Q10) Convert date/time columns to appropriate data types and decide how missing values should be handled

```python
#10
import pandas as pd
df1=pd.read_csv("customers.csv")
df3=pd.read_csv("orders.csv")
print(df1) # Dataframe of customers
print(df3) # Dataframe of orders
df1["signup_date"]=pd.to_datetime(df1["signup_date"])
df3["order_date"]=pd.to_datetime(df3["order_date"])
print(df1.dtypes) # Datatypes of customers dataframe
print(df3.dtypes) # Datatypes of orders dataframe

# Total Sum of Missing Values of Customers and orders dataframe
print(df1.isnull().sum())
print(df1.isnull().sum().sum())

print(df3.isnull().sum())
print(df3.isnull().sum().sum())

# Dropping the Missing Values of customers and orders dataframe
print(df1.dropna(inplace=True))
print(df3.dropna(inplace=True))
```

There are total 3 missing values in customers dataframe and 1 missing value in orders dataframe.
In order to handle the missing values of customers and orders,one can drop the missing values of both the dataframes
