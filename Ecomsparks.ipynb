Q11) Create a SparkSession and load all CSV files into Spark DataFrames.


```python
#11
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
sparks=SparkSession.builder.appName("Ecomspipeline").getOrCreate()
df1=sparks.read.csv("customers.csv",header=True,inferSchema=True)
df2=sparks.read.csv("order_items.csv",header=True,inferSchema=True)
df3=sparks.read.csv("orders.csv",header=True,inferSchema=True)
df4=sparks.read.csv("payments.csv",header=True,inferSchema=True)
df5=sparks.read.csv("products.csv",header=True,inferSchema=True)
# Dataframe of customers,order,order_items,products and payments
print(df1.show())
print(df2.show())
print(df3.show())
print(df4.show())
print(df5.show())
```

Q12)Apply explicit schemas to the important columns instead of relying entirely on schema inference.


```python
#12
# Apply explicit schema to important columns
from pyspark.sql import SparkSession
from pyspark.sql.types import (StructType,StructField
                               ,IntegerType,StringType,DateType)
spark=SparkSession.builder.appName("ExplicitSchema").getOrCreate()


# Concept of Explicit schema

schema1=StructType([StructField("customer_id",IntegerType(),True),
                  StructField("customer_name",StringType(),True),
                   StructField("email",StringType(),True),
                   StructField("city",StringType(),True),
                   StructField("signup_date",DateType(),True)
                  ])
df1=spark.read.csv("customers.csv",header=True,schema=schema1)
print(df1.printSchema())
print(df1.show())


schema2=StructType([StructField("order_item_id",IntegerType(),True),
                    StructField("order_id",IntegerType(),True),
                    StructField("product_id",IntegerType(),True),
                    StructField("quantity",IntegerType(),True),
                    StructField("unit_price",IntegerType(),True),
                    StructField("discount_pct",IntegerType(),True)
    ])

df2=spark.read.csv("order_items.csv",header=True,schema=schema2)
print(df2.printSchema())
print(df2.show())
```

The above mentioned are two examples of explicit schema.Rest of the schema's ie of orders,payments and products can be implemented in same way

Q14)Cast quantity, unit_price, unit_cost, and discount_pct to appropriate numeric data types.


```python
#14
from pyspark.sql.types import IntegerType,FloatType
print(df2.show()) # Dataframe of order_items
print(df5.show()) # Dataframe of products
df2=df2.withColumn("quantity",col("quantity").cast(FloatType()))
df2=df2.withColumn("unit_price",col("unit_price").cast(FloatType()))
df2=df2.withColumn("discount_pct",col("discount_pct").cast(FloatType()))
df5=df5.withColumn("unit_cost",col("unit_cost").cast(FloatType()))
print(df2.dtypes) # Datatypes of orders dataframe
print(df5.dtypes)# Datatypes of products dataframe
```

Q15) Create a line_total column using:

line_total = quantity × unit_price × (1 - discount_pct / 100)


```python
#15
df2=df2.withColumn("line_total",col("quantity")*col("unit_price")*(1-col("discount_pct")/100))
print(df2.show()) # Dataframe of order_items with Line_total column

```

Q16) Join orders, order_items, products, and customers to create a curated order-item dataset.


```python
#16
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("Datasetjoins").getOrCreate()
print(df1.show()) # Customers dataframe
print(df2.show()) # Dataframe of order_items
print(df3.show()) # Dataframe of orders
print(df5.show()) # Dataframe of products
# Creation of views of customers,orders,order_items and products
df1.createOrReplaceTempView("customers")
df2.createOrReplaceTempView("order_items")
df3.createOrReplaceTempView("orders")
df5.createOrReplaceTempView("products")

results=spark.sql('''
select * from customers c join orders o
on c.customer_id=o.customer_id
join order_items od
on o.order_id=od.order_id
join products p
on od.product_id=p.product_id

''')

print(results.show()) # Curated order item dataset after joining
```

Q17) Calculate profit and profit_margin, then write the curated dataset to an output folder in CSV or Parquet format.


```python
#17
from pyspark.sql.functions import col
df6=results
df6=df6.withColumn("profit",col("line_total")-(col("quantity")*col("unit_cost")))
df6=df6.withColumn("profit_margin",col("profit")/col("line_total")*100)
print(df6.show()) # Dataframe with profit and profit margin
# Writing the curated dataset to an output folder

df6.write\
.mode("overwrite")\
.csv("output/curated/")
```
